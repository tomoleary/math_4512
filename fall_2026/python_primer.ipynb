## 1. Variables and arithmetic

A variable gives a useful name to a value. Python evaluates the right side of `=` first, then stores the result under the name on the left.


```python
number_of_subintervals = 8
interval_length = 1.0
mesh_width = interval_length / number_of_subintervals

print("mesh width =", mesh_width)
```

Change `number_of_subintervals` to 4 or 16 and rerun the cell. Notice how `mesh_width` changes.


## 2. Arrays and indexing

A NumPy array stores several numbers together. Python begins counting at zero, so `values[0]` is the first entry.


```python
import numpy as np

number_of_subintervals = 8
grid_points = np.linspace(0.0, 1.0, number_of_subintervals + 1)
function_values = np.sin(np.pi * grid_points)

print("grid points =", grid_points)
print("first value =", function_values[0])
print("last value =", function_values[-1])
```

Try printing `function_values[3]`. Then change the sine to `np.sin(2 * np.pi * grid_points)`.


## 3. Loops

A `for` loop repeats an instruction once for each value in an array. Indentation is part of Python's syntax: the indented lines belong to the loop.


```python
mesh_width = 1.0 / number_of_subintervals
sum_of_squares = 0.0

for value in function_values:
    sum_of_squares = sum_of_squares + value**2

discrete_l2_norm = np.sqrt(mesh_width * sum_of_squares)
print("discrete L2 norm =", discrete_l2_norm)
```

The line `value**2` means \(value^2\). Try changing it to `abs(value)` and describe what is being summed instead.


## 4. A function that builds a matrix

A function groups a reusable calculation. This version of the Dirichlet discrete Laplacian uses plain arrays and an explicit loop. The row number starts at zero, even though the mathematical unknowns are often labeled \(U_1,\ldots,U_{N-1}\).


```python
def make_dirichlet_laplacian(number_of_subintervals, interval_length):
    mesh_width = interval_length / number_of_subintervals
    number_of_unknowns = number_of_subintervals - 1
    matrix = np.zeros((number_of_unknowns, number_of_unknowns))

    for row in range(number_of_unknowns):
        matrix[row, row] = 2.0
        if row > 0:
            matrix[row, row - 1] = -1.0
        if row < number_of_unknowns - 1:
            matrix[row, row + 1] = -1.0

    matrix = matrix / mesh_width**2
    return matrix

matrix = make_dirichlet_laplacian(4, 1.0)
print(matrix)
```

Read one row at a time. The diagonal receives `2`, while existing neighboring entries receive `-1`. Change 4 to 5 and identify the size of the new matrix.


## 5. Matrix-vector multiplication

The symbol `@` means matrix-vector multiplication. Here it applies the discrete approximation to \(-u''\) to a vector of three interior values.


```python
interior_values = np.array([0.1, 0.15, 0.1])
operator_values = matrix @ interior_values

print("interior values =", interior_values)
print("A_h u =", operator_values)
```

## 6. A simple plot

Use markers when plotting numerical samples. Label every axis and give the plot a descriptive title.


```python
import matplotlib.pyplot as plt

plt.plot(grid_points, function_values, "o-", label="samples of sin(pi x)")
plt.xlabel("x")
plt.ylabel("value")
plt.title("A function represented by samples")
plt.grid()
plt.legend()
plt.show()
```
